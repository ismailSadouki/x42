```python
import numpy as np
from sklearn.datasets import load_iris

# 1. CHARGEMENT ET PRÉPARATION DES DONNÉES
def prepare_data():
    iris = load_iris()
    X = iris.data  # Matrix of 150 samples x 4 features
    y = iris.target # Vector of 150 labels (0, 1, or 2)
    
    # One-Hot Encoding: converts [0, 1, 2] into [[1,0,0], [0,1,0], [0,0,1]]
    Y_true = np.eye(3)[y] 
    
    return X, Y_true, y

# 2. FONCTIONS D'ACTIVATION
def relu(z):
    return np.maximum(0, z)

def softmax(z):
    # Subtracting max(z) for numerical stability (prevents 'inf')
    exp_z = np.exp(z - np.max(z, axis=1, keepdims=True))
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)

# 3. INITIALISATION DU RÉSEAU (Poids aléatoires)
def init_network(input_size, hidden_size, output_size):
    np.random.seed(42) # For reproducible results
    network = {
        'W1': np.random.randn(input_size, hidden_size) * 0.01,
        'b1': np.zeros(hidden_size),
        'W2': np.random.randn(hidden_size, output_size) * 0.01,
        'b2': np.zeros(output_size)
    }
    return network

# 4. FORWARD PASS (Vectorized)
def forward_pass(X, net):
    # Layer 1 (Hidden)
    Z1 = np.dot(X, net['W1']) + net['b1']
    A1 = relu(Z1)
    
    # Layer 2 (Output)
    Z2 = np.dot(A1, net['W2']) + net['b2']
    Y_pred = softmax(Z2)
    
    return Y_pred

# 5. CALCUL DE LA PERTE ET DE L'ACCURACY
def compute_metrics(Y_true, Y_pred):
    # Categorical Cross-Entropy (CCE)
    # We add 1e-15 to avoid log(0)
    n_samples = Y_true.shape[0]
    loss = -np.sum(Y_true * np.log(Y_pred + 1e-15)) / n_samples
    
    # Accuracy
    pred_labels = np.argmax(Y_pred, axis=1)
    true_labels = np.argmax(Y_true, axis=1)
    accuracy = np.mean(pred_labels == true_labels) * 100
    
    return loss, accuracy

# --- EXÉCUTION DU TP ---

# Step 1: Data
X, Y_true, labels = prepare_data()

# Step 2: Network setup (4 inputs -> 10 hidden -> 3 outputs)
mlp = init_network(input_size=4, hidden_size=10, output_size=3)

# Step 3: Forward Pass
Y_probabilities = forward_pass(X, mlp)

# Step 4: Results
loss, acc = compute_metrics(Y_true, Y_probabilities)

print("-" * 30)
print(f"RESULTS FOR IRIS DATASET")
print("-" * 30)
print(f"Shape of Input Matrix X: {X.shape}")
print(f"Shape of Output Matrix (Predictions): {Y_probabilities.shape}")
print(f"Mean CCE Loss: {loss:.4f}")
print(f"Model Accuracy: {acc:.2f}%")
print("-" * 30)

# Preview of first 5 predictions
print("\nFirst 5 probability distributions (Softmax output):")
print(Y_probabilities[:5])
```
